## Q1. Accessing List Elements
numbers = [10, 20, 30, 40, 50, 60, 70]
Extract the first, middle, and last elements without modifying the original list.

```python
numbers = [10, 20, 30, 40, 50, 60, 70]
first_number = numbers[0]
middle_number = numbers[3]
last_number = numbers[6]
print(f"The First Element is {first_number}")
print(f"The Middle Element is {middle_number}")
print(f"The Last Element is {last_number}")
```

## Q2. List Slicing
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
Create three new lists: first five elements, last four elements, and every second element in reverse order.

```python
numbers2 = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
first_five_elements = numbers2[:5]
last_four_elements = numbers2[-4:]
every_second_element = numbers2[::-2]
print(f"The First Five Elements is {first_five_elements}")
print(f"The Last Four Element is {last_four_elements}")
print(f"Every Second Element is {every_second_element}")
```

## Q3. List Modification
fruits = ["apple", "banana", "cherry", "mango"]
Replace banana with orange, add grapes at index 2, and remove cherry

```python
fruits = ["apple", "banana", "cherry", "mango"]
fruits[1] = "orange"
fruits.insert(2,"grapes")
fruits.remove("cherry")
print(f"{fruits}")
```

## Q4. Reverse Without reverse()
numbers = [12, 45, 23, 67, 89, 34]
Reverse using slicing only; preserve the original list.


```python
numbers3 = [12, 45, 23, 67, 89, 34]
reverse = numbers3[::-1]
print(f"{reverse}")
```

## Q6. Rotate a List
numbers = [10, 20, 30, 40, 50, 60]
k = 2
Rotate the list right by k using slicing. Expected: [50, 60, 10, 20, 30, 40].

```python
numbers5 = [10, 20, 30, 40, 50, 60]
k = 2
rotate_list = numbers5[-k:]
rotate_list1 = numbers5[:-k]
print(f"{rotate_list+rotate_list1}")
```

## Q9. Extract Data from a Matrix
matrix = [
 [10, 20, 30],
 [40, 50, 60],
 [70, 80, 90]
]
Extract main diagonal, secondary diagonal, and middle column; calculate the sum of each.


```python
matrix = [
 [10, 20, 30],
 [40, 50, 60],
 [70, 80, 90]
]

main_diagonal = matrix[0][0]+matrix[1][1]+matrix[2][2]
secondary_diagonal = matrix[0][2]+matrix[1][1]+matrix[2][0]
middle_column = matrix[0][1]+matrix[1][1]+matrix[2][1]
print(f"The Main Diagonal is {main_diagonal}")
print(f"The Secondary Diagonal is {secondary_diagonal}")
print(f"The Middle Column is {middle_column}")
```

## Q10. List Transformation Challenge
data = [15, 5, 25, 10, 35, 20, 30]
Build a list of the three largest unique values ascending, then two smallest unique values descending. Expected: [25,30, 35, 10, 5]

```python
data1 = [15, 5, 25, 10, 35, 20, 30]
data1.sort()
three_largest_unique = data1[4:]
two_smallest_unique = data1[:2]
two_smallest_unique.sort(reverse=True)
print(f"{three_largest_unique + two_smallest_unique}")
print()


```
