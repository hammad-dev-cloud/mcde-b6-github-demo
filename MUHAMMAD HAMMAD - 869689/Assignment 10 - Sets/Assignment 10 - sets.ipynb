## Q22. Set Operations

Find union, intersection, a - b, and symmetric difference.

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}
Uninon = a | b
Intersection = a & b
Subtract = a-b
difference = a^b
print(f"The Set of Union is {Uninon}")
print(f"The Set of Intersection is{Intersection}")
print(f"The Set of Subtract is {Subtract}")
print(f"The Set of Symmetric Difference is {difference}")
```

## Q27. Three-Set Analysis
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}
Find students in all three, exactly Python and SQL (not AWS), and at least two

```python
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

all_three = python & sql & aws
exactly_python_sql = (python & sql) - aws
atleast_two = (python & sql) | (python & aws) | (aws & sql)


print(f"All three {all_three}")
print(f"Exactly Python and Sql Not Aws {exactly_python_sql}")
print(f"At least two {atleast_two}")
```

## Q28. Transaction Analysis
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}
Find IDs appearing on all days, exactly one day, and at least two days.

```python
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}


on_all_days = day1 & day2 & day3
at_least_two_day = (day1 & day2) | (day1 & day3) | (day2 & day3)
exactly_one_day = (day1 | day2 | day3) - (at_least_two_day)

print (f"On All Day {on_all_days}")
print(f"At Least Two Day {at_least_two_day}")
print(f"Exatly One Day {exactly_one_day}")

```

## Q29. Access Control
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}
Find missing and extra permissions and test full authorization.

```python
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

mp = required_permissions - user_permissions
ep = user_permissions - required_permissions

print(f"Missing Permission {mp}")
print(f"Extra Permission {ep}")
print(f"Full Authorization {user_permissions.issuperset(required_permissions)}")
```

## Q30. Customer Segmentation
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}
Find customers on exactly one, exactly two, and all three platforms.

```python
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

at = website & mobile & store
et = ((website & mobile)|(website & store)|(mobile & store)) - (at)
eo = (website|mobile|store) - et - at

print(f"All Three Platforms {at}")
print(f"Exactly Two Platforms {et}")
print(f"Exactly One Platforms {eo}")
```
