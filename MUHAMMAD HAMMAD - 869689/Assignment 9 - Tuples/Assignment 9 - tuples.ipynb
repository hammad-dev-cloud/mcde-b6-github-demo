## Q17. Nested Tuple Access
students = (
 ("Ali", (85, 90, 88)),
 ("Sara", (92, 87, 95)),
 ("Ahmed", (78, 84, 80))
)
Extract Sara second score, Ali average, and Ahmed highest score.

```python
students = (
 ("Ali", (85, 90, 88)),
 ("Sara", (92, 87, 95)),
 ("Ahmed", (78, 84, 80))
)

s1 = sum(students[0][1])/3
s2 = (students[1][1][1])
s3 = max(students[2][1])
print(f"Ali Average Score is {s1}")
print(f"Sara Second Score is {s2}")
print(f"Ahmed Highest Score is {s3}")
```

## Q19. Tuple Matrix
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))
Extract diagonal, calculate its product, and build a tuple of row sums.

```python
matrix = ((2, 4, 6),
          (8, 10, 12),
          (14, 16, 18))

p1 = (matrix[0][0]* matrix[1][1]* matrix[2][2])
d1 = (matrix[0][0], matrix[1][1], matrix[2][2])
s1 = sum(matrix[0])
s2 = sum(matrix[1])
s3 = sum(matrix[2])
print(f"Diagonal is {d1}")
print(f"Product is {p1}")
print(f"Rows Sums is {s1,s2,s3}")
```

## Q20. Inventory Tuple Challenge
records = (
 ("Laptop", 120000, 5),
 ("Mouse", 1500, 20),
 ("Keyboard", 3500, 10)
)
Calculate inventory value, find most expensive product, and build reversed tuple of names.

```python
records = (
 ("Laptop", 120000, 5),
 ("Mouse", 1500, 20),
 ("Keyboard", 3500, 10)
)
c1 = (records[0][1]*records[0][2])
c2 = (records[1][1]*records[1][2])
c3 = (records[2][1]*records[2][2])
cv = (c1+c2+c3)
mp = max(records[0][1],records[1][1],records[2][1])
if mp == records[0][1]:
    ep = records[0][0]
elif mp == records[1][1]:
    ep = records[1][0]
else :
    ep = records[2][0]
br = (records[2][0],records[1][0],records[0][0])
print(f"Inventory Value is {cv}")
print(f"Expensive Product Name is {ep}")
print (f"Reversed Tuple is {br}")
```
