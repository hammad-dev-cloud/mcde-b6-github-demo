## Q35. Student Marks
marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}
Find highest, lowest, average, and top-scoring student without loops or lambda.

```python
marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}

highest = max(marks.values())
lowest = min(marks.values())
count= len(marks)
top_scoring = max(marks, key=marks.get)
average = sum(marks.values())/count

print(f"Highest Marks is {highest}")
print(f"Lowest Makrs is {lowest}")
print(f"Average Marks is {average}")
print(f"Top Scoring is {top_scoring}")
```

## Q36. Nested Employee Records

Extract Sara salary, increase Ali salary 15%, and calculate total payroll.

## Q38. Order Calculation
order = {
 "laptop": {"price": 120000, "quantity": 2},
 "mouse": {"price": 1500, "quantity": 5},
 "keyboard": {"price": 3500, "quantity": 3}
}
Calculate subtotals, grand total, and amount after 10% discount

```python
order = {
 "laptop": {"price": 120000, "quantity": 2},
 "mouse": {"price": 1500, "quantity": 5},
 "keyboard": {"price": 3500, "quantity": 3}
}
laptop_subtotal1 = order["laptop"]["price"]* order["laptop"]["quantity"]
mouse_subtotal2 = order["mouse"]["price"]* order["mouse"]["quantity"]
keyboard_subtotal3 = order["keyboard"]["price"]* order["keyboard"]["quantity"]

grand_total = laptop_subtotal1 + mouse_subtotal2 + keyboard_subtotal3
discount = grand_total*10/100
after_discount = grand_total-discount

print(f"Grand Total is {grand_total}")
print(f"Amount After Discount is {after_discount}")
print(f"Laptop Subtotal is {laptop_subtotal1}")
print(f"Mouse Subtotal is {mouse_subtotal2}")
print(f"Keyboard Subtotal is {keyboard_subtotal3}")
```

## Q39. Employee Performance
employees = {
 "Ali": {"python": 90, "sql": 85, "aws": 88},
 "Sara": {"python": 95, "sql": 92, "aws": 90},
 "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}
Calculate each average, top performer, and highest SQL score

```python
employees = {
 "Ali": {"python": 90, "sql": 85, "aws": 88},
 "Sara": {"python": 95, "sql": 92, "aws": 90},
 "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

ali_avg = (employees["Ali"]["python"]+
            employees["Ali"]["sql"]+
            employees["Ali"]["aws"]
            )/3
         
sara_avg = (employees["Sara"]["python"]+
            employees["Sara"]["sql"]+
            employees["Sara"]["aws"]
            )/3            

ahmed_avg = (employees["Ahmed"]["python"]+
            employees["Ahmed"]["sql"]+
            employees["Ahmed"]["aws"]
            )/3

highest = max(employees["Ali"]["sql"] , employees["Sara"]["sql"] , employees["Ahmed"]["sql"])

top = max(ali_avg,sara_avg,ahmed_avg)

print(f"Ali Average {ali_avg}")
print(f"Sara Average {sara_avg}")
print(f"Ahmed Average {ahmed_avg}")

print(f"Highest in SQL Score {highest}")


if top == ali_avg:
    print(f"Ali is Top in Employee Performance {ali_avg}")
elif top == sara_avg:
    print(f"Sara is Top in Employee Performance {sara_avg}")
else:
    print(f"Ahmed is Top in Employee Performance {ahmed_avg}")
```

## Q40. Inventory Management
inventory = {
 "P101": {"name": "Laptop", "price": 120000, "stock": 5},
 "P102": {"name": "Mouse", "price": 1500, "stock": 20},
 "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}
Calculate total value, most valuable item by price x stock, and update stock after selling two laptops and five mice.

```python
inventory = {
 "P101": {"name": "Laptop", "price": 120000, "stock": 5},
 "P102": {"name": "Mouse", "price": 1500, "stock": 20},
 "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

laptop = inventory["P101"]["price"]* inventory["P101"]["stock"]
mouse = inventory["P102"]["price"]* inventory["P102"]["stock"]
keyboard = inventory["P103"]["price"]* inventory["P103"]["stock"]

mv_item = max(laptop,mouse,keyboard)


total = laptop + mouse + keyboard

inventory["P101"]["stock"]-2
inventory["P102"]["stock"]-5

print(f"Total Value {total}")

if mv_item == laptop:
    print(f"Most Valueable Item Laptop is {laptop}") 
elif mv_item == mouse:
    print(f"Most Valueable Item Mouse is {mouse}") 
else:
    print(f"Most Valueable Item Keyboard is {keyboard}")

print(f"Laptop Stock: {inventory['P101']['stock']}")
print(f"Mouse Stock: {inventory['P102']['stock']}")



```
