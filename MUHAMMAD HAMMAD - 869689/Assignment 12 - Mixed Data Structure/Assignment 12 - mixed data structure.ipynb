## Q45. Course Enrollment
courses = {
 "Python": {"Ali", "Sara", "Ahmed"},
 "SQL": {"Sara", "Ahmed", "Zain"},
 "AWS": {"Ali", "Ahmed", "Zain"}
}
Find students in exactly two, all three, and exactly one course; count unique students.


```python

courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

python_sql = courses["Python"] & courses["SQL"]
python_aws = courses["Python"] & courses["AWS"]
sql_aws = courses["SQL"] & courses["AWS"]

all_three = courses["Python"] & courses["SQL"] & courses["AWS"]

exactly_two = (python_sql | python_aws | sql_aws) - all_three

all_students = courses["Python"] | courses["SQL"] | courses["AWS"]

exactly_one = all_students - (python_sql | python_aws | sql_aws)

unique_count = len(all_students)

print(f"Students in Exactly Two Courses: {exactly_two}")
print(f"Students in All Three Courses: {all_three}")
print(f"Students in Exactly One Course: {exactly_one}")
print(f"Count of Unique Students: {unique_count}")
```

pipelines = {
 "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},

}
Find total processed records, failed pipelines, largest pipeline total, and average records per run for each.

```python

pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

p1 = sum(pipelines["pipeline_1"]["records"])
p2 = sum(pipelines["pipeline_2"]["records"])
p3 = sum(pipelines["pipeline_3"]["records"])

total = p1 + p2 + p3

avg1 = p1 / len(pipelines["pipeline_1"]["records"])
avg2 = p2 / len(pipelines["pipeline_2"]["records"])
avg3 = p3 / len(pipelines["pipeline_3"]["records"])

largest = max(p1, p2, p3)

failed = set()

if pipelines["pipeline_1"]["status"] == "failed":
    failed.add("pipeline_1")

if pipelines["pipeline_2"]["status"] == "failed":
    failed.add("pipeline_2")

if pipelines["pipeline_3"]["status"] == "failed":
    failed.add("pipeline_3")

print(f"Pipeline 1 Total: {p1}")
print(f"Pipeline 2 Total: {p2}")
print(f"Pipeline 3 Total: {p3}")
print(f"Total Processed Records: {total}")
print(f"Failed Pipelines: {failed}")

if largest == p1:
    print(f"Largest Pipeline: pipeline_1 - {largest}")
elif largest == p2:
    print(f"Largest Pipeline: pipeline_2 - {largest}")
else:
    print(f"Largest Pipeline: pipeline_3 - {largest}")

print(f"Pipeline 1 Average Records: {avg1}")
print(f"Pipeline 2 Average Records: {avg2}")
print(f"Pipeline 3 Average Records: {avg3}")
```

Find products bought by at least two users, Ali wishlist minus purchases, shared wishlist products, and all products.

## Q50. Final Company Analysis
company = {
 "departments": {
 "Engineering": {
 "employees": [
 ("Ali", 120000, {"Python", "AWS", "SQL"}),
 ("Sara", 150000, {"Python", "Databricks", "SQL"})
 ],
 "budget": 500000
 },
 "Analytics": {
 "employees": [
 ("Ahmed", 100000, {"SQL", "Power BI"}),
 ("Zain", 130000, {"Python", "SQL", "Tableau"})
 ],
 "budget": 400000
 }
 }
}
1. Payroll per department. 2. Company payroll. 3. Highest-paid employee. 4. Skills shared by all. 5. Engineering-only
skills. 6. Remaining budget per department. 7. Department with higher average salary. 8. Name-to-salary dictionary. 9.All unique skills. 10. Is combined budget sufficient?

```python

company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

eng = company["departments"]["Engineering"]
ana = company["departments"]["Analytics"]

eng_payroll = eng["employees"][0][1] + eng["employees"][1][1]
ana_payroll = ana["employees"][0][1] + ana["employees"][1][1]

print(f"Engineering Payroll: {eng_payroll}")
print(f"Analytics Payroll: {ana_payroll}")

company_payroll = eng_payroll + ana_payroll

print(f"Company Payroll: {company_payroll}")

salaries = (
    eng["employees"][0][1],
    eng["employees"][1][1],
    ana["employees"][0][1],
    ana["employees"][1][1]
)

highest = max(salaries)

if highest == eng["employees"][0][1]:
    print(f"Highest Paid Employee: {eng['employees'][0][0]} - {highest}")
elif highest == eng["employees"][1][1]:
    print(f"Highest Paid Employee: {eng['employees'][1][0]} - {highest}")
elif highest == ana["employees"][0][1]:
    print(f"Highest Paid Employee: {ana['employees'][0][0]} - {highest}")
else:
    print(f"Highest Paid Employee: {ana['employees'][1][0]} - {highest}")

shared_skills = (
    eng["employees"][0][2]
    & eng["employees"][1][2]
    & ana["employees"][0][2]
    & ana["employees"][1][2]
)

print(f"Skills Shared by All: {shared_skills}")

eng_skills = eng["employees"][0][2] | eng["employees"][1][2]
ana_skills = ana["employees"][0][2] | ana["employees"][1][2]

engineering_only = eng_skills - ana_skills

print(f"Engineering Only Skills: {engineering_only}")

eng_remaining = eng["budget"] - eng_payroll
ana_remaining = ana["budget"] - ana_payroll

print(f"Engineering Remaining Budget: {eng_remaining}")
print(f"Analytics Remaining Budget: {ana_remaining}")

eng_avg = eng_payroll / len(eng["employees"])
ana_avg = ana_payroll / len(ana["employees"])

if eng_avg > ana_avg:
    print(f"Higher Average Salary: Engineering - {eng_avg}")
elif ana_avg > eng_avg:
    print(f"Higher Average Salary: Analytics - {ana_avg}")
else:
    print(f"Both Departments Have Equal Average Salary: {eng_avg}")

salary_dict = {
    eng["employees"][0][0]: eng["employees"][0][1],
    eng["employees"][1][0]: eng["employees"][1][1],
    ana["employees"][0][0]: ana["employees"][0][1],
    ana["employees"][1][0]: ana["employees"][1][1]
}

print(f"Name to Salary Dictionary: {salary_dict}")

all_skills = eng_skills | ana_skills

print(f"All Unique Skills: {all_skills}")

total_budget = eng["budget"] + ana["budget"]

if total_budget >= company_payroll:
    print(f"Combined Budget Sufficient: Yes")
else:
    print(f"Combined Budget Sufficient: No")
```
